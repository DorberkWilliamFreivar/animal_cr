```python
import pandas as pds;
import numpy as npy;
import os as kindos;
```

```python
def customize_tabel(data):
    res = data.style.set_table_styles([
        {'selector': 'thead th', 
        'props': [('background-color', "#DF5ED4"), ('color', 'white'), ('font-weight', 'bold'), ('border', '1px solid white'), ('text-align', 'left')]},

        {'selector': 'tbody td', 
        'props': [('background-color', '#202124'), ('color', 'white'), ('border', '1px solid white'), ('text-align', 'left')]}
    ]).hide(axis='index');
    return res;

def get_file(target_file):
    BASE_DIR = kindos.getcwd();
    for curr_folder, list_folders, list_files in kindos.walk(BASE_DIR):
        if target_file in list_files:
            return kindos.path.join(curr_folder, target_file);
    return None;
```
